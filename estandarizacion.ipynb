**Estandarización:** Convertir los datos de entrada para que tengan media 0 y desviación estándar 1, a este proceso le llamamos centrado y escalado.

**Fórmula:**

$X_{ij}^{S}=\left\{\frac{X_{ij}-\operatorname{media}(X_i)}{\operatorname{desv}(X_i)},\ \forall j\in 1,\ldots,n\right\}
$

```python
from sklearn import preprocessing
import numpy as np
```

**Preprocessing:** Se utiliza para el procesamiento de datos: limpieza, tranformación y escalado antes de entrenar un modelo de Machine Learning.

**Numpy:** Se utiliza para trabajar con vecrtores y matrices, realizar operaciones numericas y manejar arreglos multidimensionales de forma eficiente. 

```python
# Carga de datos.
# Columna 1: peso en kg, Columna 2: altura en metros, Columna 3: edad en años
X_train = np.array([[70., 1.65, 20.],
                    [80., 1.75, 25.],
                    [60., 1.60, 30.]])
print(X_train)
```

```python
# Limpieza de datos: estandarización.
standardizer = preprocessing.StandardScaler()
X_train_std = standardizer.fit_transform(X_train)
print(X_train_std)
```

**StandardScaler:** Se utiliza para sacar la media y la desviación estándar de un conjunto de datos.

**Fit:** Calcula y almacena la media y la desviación estándar del conjunto de datos.

**Transform:** Transforma los valores calculados para que queden con media 0 y desviación estándar 1.

```python
# Salida de resultados.
print(X_train_std)
```

```python
# Ejemplo para ver que los resultados son iguales si aplicamos los métodos "fit" y "transform" por separado
X_train_2 = np.copy(X_train)
```

```python
# Aplicamos los métodos "fit" y "transform" por separado
f = preprocessing.StandardScaler()
f = f.fit(X_train)
X_train_2_std = f.transform(X_train_2)
print(X_train_2_std)
```

```python
# Comprobar si las matrices estandarizadas son iguales
print(np.array_equiv(X_train_std, X_train_2_std))
```
