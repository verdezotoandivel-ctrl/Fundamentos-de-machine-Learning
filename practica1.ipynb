```python
from sklearn import preprocessing
import numpy as np
```

**MinMaxScaler:** Función que permite encontrar los valores minimos y maximos de un conjunto de datos.

**Numpy:** 

```python
# Carga de datos.
X_train = np.array([[ 1., -4., 5.],
[ 5., 0., 2.],
[ 0., 6., -3.]])
print(X_train)
```

```python
# Limpieza de datos: normalización.
normalizer = preprocessing.MinMaxScaler()
X_train_norm = normalizer.fit_transform(X_train)
```

**Fit:** Entrena.

**Transform:** Transforma.

```python
# Salida de resultados.
print(X_train_norm)
```

```python
# Ejemplo para ver que los resultados son iguales si aplicamos los métodos␣"fit" y "transform" por separado
X_train_2 = np.copy(X_train)

# Crea un objeto "fitted_normalizer" llamando al método "fit"
f = normalizer.fit(X_train)

# Aplica el objeto sobre los datos "X_train_2" para obtener "X_train_norm_2"
# Imprimo la matriz x_train-norm_2
X_train_norm_2 = f.transform(X_train)

# Imprimo la matriz x_train-norm_2
# Comprueba si "X_train_norm" y "X_train_norm_2" son iguales
print(np.array_equiv(X_train_norm, X_train_norm_2))
```

```python
# Normalizar un nuevo conjunto de datos de test (SOLO HAY QUE APLICAR EL MÉTODO␣"transform", NO EL "fit")
X_test = np.array([[ 1., -3., 2.],
[ 3., 0., 0.],
[ 0., 4., -1.]])

# Normalizar los datos de "X_test"
#Un solo entrenaiento para encontrar 
X_test_norm=f.transform(X_test)
```

```python
from sklearn import preprocessing
import numpy as np
```

```python
# Carga de datos con nuevos valores (Ej: [Peso kg, Altura , Edad])
X_train = np.array([[ 70, 175, 25],
                    [ 85, 180, 40],
                    [ 52, 160, 19],
                    [ 90, 168, 55]])

X_test = np.array([[ 65, 170, 30],
                   [ 80, 185, 48]])
```

```python
# Inicializamos el estandarizador
standardizer = preprocessing.StandardScaler()

# Estandarizamos los datos de entrenamiento
X_train_std = standardizer.fit_transform(X_train)
print("X_train estandarizado:\n", X_train_std)
```

```python
# Aplicamos la transformación al conjunto de prueba usando los parámetros del entrenamiento
X_test_std = standardizer.transform(X_test)
print("X_test estandarizado:\n", X_test_std)
```

```python
# Aplicamos los métodos "fit" y "transform" por separado
f = preprocessing.StandardScaler()
f = f.fit(X_train)

X_train_2_std = f.transform(X_train)
X_test_2_std = f.transform(X_test)

print("X_train_2_std:\n", X_train_2_std)
```
